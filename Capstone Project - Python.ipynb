## Task 2

The cost column in the dataframe “Project” has some missing values. Your task is to 
compute these missing values. Replace the missing values by running average. You 
should use the “For” loop for this task.

```python
import pandas as pd
import numpy as np

Project = pd.read_csv('Data/project.csv') #Loading the already saved project dataframe
```

```python
Project
```

```python
running_sum = 0 #Initializing running sum to 0

running_count = 0 #initializing running count to zero. These variables help in calculating running average

for i in range(len(Project)): #defining for loop over the length of the dataframe
    if pd.isnull(Project.loc[i,'Cost']):   #if condition when the loop iterates over the cell that is nan
        Project.loc[i,'Cost'] = running_sum/running_count #assigns the running average to the nan cell
        running_sum += Project.loc[i, 'Cost'] #The same value is added to the running sum
        running_count += 1 #providing increment to the count of rows for computing average
    else:      #else condition when loop iterates over a cell that is not nan
        running_sum += Project.loc[i,'Cost'] #Value in the current cell added to running sum
        running_count += 1 #providing increment to the count of rows for computing average

Project['Cost'] = Project['Cost'].astype(int) #changing the datatype of the 'Cost' column to integer

Project
```

```python
Project.to_csv('Project_Task2.csv', index=False) #saving the updates made to the Project dataframe
```

## Task 6

Demote the designation level by 1, whose projects have status “fail”. Delete the 
employees record whose designation level is above 4.

```python
import numpy as np
import pandas as pd

Final = pd.read_csv("Final_Task5.csv") #Loading the already saved Final dataframe
```

```python
failed_ids = Final[Final["Status"] == "Failed"]["ID"].unique() #identifying all IDs which have failed projects

Final["Designation Level"] = np.where( #Demoting the designation level by 1 where any project has status "Failed".
    Final["ID"].isin(failed_ids),     #Designation level is to be treated as a property of the employee
    Final["Designation Level"] - 1,   #So changes in designation level should apply consistently across all projects
    Final["Designation Level"]
)

Final = Final[Final["Designation Level"] < 4] #Deleting records where the designation level is above 4

Final
```

```python
Final.to_csv("Final_Task6.csv", index = False) #Saving the updates to Final dataframe
```

## Task 9

Add the cost of all projects for each Employee and save it in new dataframe

```python
import pandas as pd

Final = pd.read_csv("Final_Task8.csv") #loading the already saved final dataframe
```

## Task 10

Print all the employee details whose city name contains the letter “o” in it.

```python
import pandas as pd

Employee = pd.read_csv("Employee_Task3.csv") #Loading the already saved Employee dataframe
TotalProjCost = pd.read_csv("TotalProjCost.csv") #Loading the already saved TotalProjCost dataframe
Final = pd.read_csv("Final_Task8.csv")##Loading the already saved Final dataframe
```

```python
city_o = Employee[Employee["City"].str.contains("o")] #identifying the employees with letter "o" in the city
city_o = city_o.drop(columns=["First Name"]) #dropping firstname column from employees since this will be taken from TotalProjCost

merged = pd.merge(city_o, TotalProjCost, on="ID", how= "left") #merging with TotalProjCost to get the total project cost for each employee

designation = Final.groupby("ID")["Designation Level"].first().reset_index() #Getting the current designation level of employees

project_count = Final.groupby("ID")["Project"].count().reset_index() #Getting the total number of projects for each selected employee
project_count = project_count.rename(columns={"Project": "Number of Projects"}) 

merged = pd.merge(merged, designation, on ="ID", how = "left") #joining the designation column

merged = pd.merge(merged, project_count, on ="ID", how = "left")

cols = merged.columns.tolist() #Reorder columns

cols.insert(1, cols.pop(cols.index("First Name"))) #move first name column to index 1

Employee_details_o = merged[cols] #reorder dataframe
Employee_details_o
```

```python
Employee_details_o.to_csv("Employee_details_o.csv", index = False)
```
